# 02 · Feature extraction (the long step)

Every image is passed once through each frozen encoder and the feature vector is saved in `features/<model>/`.
All later steps reuse these files, so this is done only once.

* One cell per model — run them one at a time, whenever you have time (e.g. one or two per evening).
* Times are for the 4 GB laptop GPU. With **Option A** (cached features) each cell finishes in seconds.
* If a cell stops, run it again: it resumes from the last saved chunk.
* Before starting: charger plugged in, sleep set to *Never*, other heavy programs closed.
* `--workers 2` loads images in parallel. If you get a *DataLoader worker* error, change it to `--workers 0`.

```python
import sys
from pathlib import Path
here = Path.cwd()
sys.path.insert(0, str(here if (here / "nbtools.py").exists() else here / "notebooks"))
from nbtools import run, show, figure, where, ROOT
where()
```
